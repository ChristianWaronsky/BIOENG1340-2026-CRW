```python
import SimpleITK as sitk
imagePath = '/content/pigdicom/'

print("Reading Dicom directory:", imagePath)
reader = sitk.ImageSeriesReader()

dicom_names = reader.GetGDCMSeriesFileNames(imagePath)
reader.SetFileNames(dicom_names)

image = reader.Execute()
```

```python
image.GetSize()
```

```python
image.GetDimension()
```

```python
image.GetSpacing()
```

```python
imageArray = sitk.GetArrayFromImage(image)

# Lets do some image processing:  Add a scalar value to every voxel's existing scalar value:
imageArray = imageArray + 10

imageNew = sitk.GetImageFromArray(imageArray)

#Copy back the meta-information pertaining to spacing of voxels to the processed image before writing it back to a file on google drive
imageNew.SetSpacing(image.GetSpacing())
sitk.WriteImage(imageNew, '/content/drive/MyDrive/1340/Week01/MyNewImage.nii.gz')
```
